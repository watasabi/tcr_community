# Testes de associação bi-variados — TCR/UTI

Desfecho padronizado (Óbito / Alta / Transferência) vs preditoras clínicas.


```python
from pathlib import Path

import pandas as pd

from tcr_community.pipeline.prepare_data import PROCESSED_PATH, prepare_data
from tcr_community.schemas.columns import (
    ASSOCIATION_PREDICTORS,
    COL_APACHE_II,
    COL_DESFECHO_PADRONIZADO,
)
from tcr_community.stats.association import run_association_battery

if not Path(PROCESSED_PATH).exists():
    prepare_data()

df = pd.read_parquet(PROCESSED_PATH)
df[COL_DESFECHO_PADRONIZADO].value_counts()
```

```python
numeric_preds = [COL_APACHE_II]
categorical_preds = [p for p in ASSOCIATION_PREDICTORS if p != COL_APACHE_II]

results = run_association_battery(
    df,
    predictors=ASSOCIATION_PREDICTORS,
    target=COL_DESFECHO_PADRONIZADO,
    numeric_predictors=numeric_preds,
)
results.sort_values("p_value")
```

```python
out = Path("../../reports/association_results.csv")
out.parent.mkdir(parents=True, exist_ok=True)
results.to_csv(out, index=False)
print(f"Salvo em {out.resolve()}")
```
