# Estatística descritiva — TCR/UTI

Tabelas de frequência (categóricas) e medidas de tendência/dispersão (quantitativas).


```python
from pathlib import Path

import pandas as pd

from tcr_community.pipeline.prepare_data import PROCESSED_PATH, prepare_data
from tcr_community.stats.descriptive import (
    clinical_report,
    demographic_report,
    export_report,
)

if not Path(PROCESSED_PATH).exists():
    prepare_data()

df = pd.read_parquet(PROCESSED_PATH)
print(f"Pacientes: {len(df)}")
df.head()
```

## Demográfica


```python
demo = demographic_report(df)
for name, table in demo.items():
    print(f"\n### {name}")
    print(table.to_string(index=False))
```

## Clínica


```python
clinical = clinical_report(df)
for name, table in clinical.items():
    print(f"\n### {name}")
    print(table.to_string(index=False))
```

```python
reports_dir = Path("../../reports/descriptive")
export_report(demo, reports_dir, prefix="demo_")
export_report(clinical, reports_dir, prefix="clin_")
print(f"Exportado em {reports_dir.resolve()}")
```
